# **CAT EXAMPLE**

**CONTEXT 1**

```python
# create a class
class Cat:
    name = 'Calico'
    age = 0.8

#test: create an object
cat = Cat()
print(cat.name)
print(cat.age)
```

What you're seeing in that code example is a basic teaching demonstration, which makes them look identical because hardcoded values are being used.

Here is the distinction:
- Class (Cat): The blueprint or cookie cutter. It defines what attributes or behaviors a cat should have.

- Object (cat): The actual instance created from that blueprint—like a cookie stamped out by the cutter.

In the screenshot, the class hardcodes the values directly. Because name and age are hardcoded inside the class, every single object you create from this class will be forced to be a 0.8-year-old Calico.

This is called a class attribute, and while it's valid Python syntax, it's rarely how classes are used in real-world code.

In real code, a class acts as a flexible template so that every individual object can hold its own unique data. To do this, Python uses a special method called ("__init__")

**CONTEXT 2**

```python
# create a class
class Cat:
    name = 'unknown'
    age = -1

#test: create an object
cat = Cat()
print(cat.name)
print(cat.age)
```

```python
# create a class
class Cat:
    name = 'unknown'
    age = -1

    def set_init_values(input_name, input_age):
        name = input_name
        age = input_age

#test: create an object
cat = Cat()
cat.set_init_values('Mimi', 0.5)
print(cat.name)
print(cat.age)
```

This happens because Python automatically passes the object instance itself (cat) as the first argument whenever you call a method on an instance.

Here is why Python says 3 arguments were given:
- cat (automatically passed by Python as the first argument)
- 'Mimi' (the 2nd argument you passed)
- 0.5 (the 3rd argument you passed)

Because your definition def set_init_values(input_name, input_age): only accepts 2 parameters, Python throws a TypeError.

In Python class methods, you need to add self as the first parameter to accept that automatic reference to the instance. You also need to attach variables to self (self.name, self.age) so they update the object's properties:

```python
# create a class
class Cat:
    name = 'unknown'
    age = -1

    def set_init_values(self, input_name, input_age):
        name = input_name
        age = input_age

#test: create an object
cat = Cat()
cat.set_init_values('Mimi', 0.5)
print(cat.name)
print(cat.age)
```

```python
# create a class
class Cat:
    name = 'unknown'
    age = -1

    def set_init_values(self, name, age):
        self.name = name
        self.age = age

#test: create an object
cat = Cat()
cat.set_init_values('Mimi', 7)
print(cat.name)
print(cat.age)
```

Parameter Catching (self): You added self as the first parameter in def set_init_values(self, name, age):

Now, when Python automatically passes cat as the first hidden argument, self catches it so the argument counts line up (3 given, 3 expected).

Variable Binding (self.): You attached self. to the variables inside the method (self.name = name).

What is the point of self.name and self.age?
Without self., Python creates temporary local variables inside the function that vanish as soon as the method finishes running.
- name (on the right) is just a local variable containing 'Mimi'.
- self.name (on the left) means: "Save this onto the actual cat object permanently."

Think of self as a pointer saying "THIS specific cat's attributes." self.name = name tells Python: "Take the incoming input value name and permanently assign it to THIS cat's name property."

```python
#create a class

class Cat: 
    def set_init_values(self, name, age):
        self.name = name
        self.age = age

#test: create an object
cat = Cat()
cat.set_init_values('Calico', 0.5)
print(cat.name)
print(cat.age)
```
