Because your definition def set_init_values(input_name, input_age): only accepts 2 parameters, Python throws a TypeError.

In Python class methods, you need to add self as the first parameter to accept that automatic reference to the instance. You also need to attach variables to self (self.name, self.age) so they update the object's properties:

```python
# create a class
class Cat:
    name = 'unknown'
    age = -1

    def set_init_values(self, input_name, input_age):
        name = input_name
        age = input_age

#test: create an object
cat = Cat()
cat.set_init_values('Mimi', 0.5)
print(cat.name)
print(cat.age)
```

```python
# create a class
class Cat:
    name = 'unknown'
    age = -1

    def set_init_values(self, name, age):
        self.name = name
        self.age = age

#test: create an object
cat = Cat()
cat.set_init_values('Mimi', 7)
print(cat.name)
print(cat.age)
```

Parameter Catching (self): You added self as the first parameter in def set_init_values(self, name, age):

Now, when Python automatically passes cat as the first hidden argument, self catches it so the argument counts line up (3 given, 3 expected).

Variable Binding (self.): You attached self. to the variables inside the method (self.name = name).

What is the point of self.name and self.age?
Without self., Python creates temporary local variables inside the function that vanish as soon as the method finishes running.
- name (on the right) is just a local variable containing 'Mimi'.
- self.name (on the left) means: "Save this onto the actual cat object permanently."

Think of self as a pointer saying "THIS specific cat's attributes." self.name = name tells Python: "Take the incoming input value name and permanently assign it to THIS cat's name property."

```python
#create a class

class Cat: 
    def set_init_values(self, name, age):
        self.name = name
        self.age = age

#test: create an object
cat = Cat()
cat.set_init_values('Calico', 0.5)
print(cat.name)
print(cat.age)
```

Here, you removed the top class-level variables (name = 'unknown' and age = -1), and the code works completely fine.

In Python, you don't need to declare attributes at the class level beforehand if you plan to attach them inside a method using self.

In the previous one: 
- name = 'unknown' was a class attribute (a default fallback value shared across the class definition).

In this:
- By calling self.name = name inside set_init_values(), Python dynamically creates an instance attribute named name directly on the cat object on the fly.

Dynamic Creation: Python allows attributes to be created on an object anytime you assign something to self.variable_name.

**CONTEXT 3**

```python
#create a class

class Cat:
    def set_init_values(self, name, age):
        self.name = name
        self.age = age

    def increase_age(self, age):
        self.age = self.age + age

    def describe(self):
        print(f'[- Name: {self.name} - Age: {self.age}]')


#test: create an object
cat = Cat()
cat.set_init_values('Miko', 5)
cat.describe()

cat.increase_age(0.5)
cat.describe()
```

```python
#create a class

class Cat:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def increase_age(self, age):
        self.age += age

    def __call__(self):
        print(f'Name: {self.name} - Age: {self.age}')

#test: create an object
cat = Cat('Momo', 9)
cat()

cat.increase_age(0.3)
cat()
```

Here are the 3 main differences between the two approaches:

1. Object Creation (set_init_values vs. __init__)
- First one: You create an "empty" cat first (cat = Cat()), and then you have to manually call a custom helper function (cat.set_init_values('Miko', 5)) to set its name and age.
- Second one: You use __init__, which is Python's standard constructor. Python automatically triggers this the instant you write Cat('Momo', 9). This allows you to initialize the object in a single line.

2. Printing/Describing (describe() vs. __call__)
- First one: You define a regular method named describe(), which requires an explicit method call: cat.describe().
- Second one: You use the special dunder method __call__(). This lets you call the object instance directly like a function: cat().

**Summary**
The second screenshot uses Python's special "dunder" (double underscore) methods (__init__ and __call__) to make the code more concise and pythonic.

**CONTEXT 4**

```python
#create a class
class Cat:
    def __init__(self, name, age):
        self.name = name
        self. age = age

    def __call__(self):
        print(f"Name: {self.name} - Age: {self.age}")

#test: create an object
cat = Cat("Momo", 4)
cat()

cat.age = cat.age + 0.3
cat()
```

```python
#create a class
class Cat:
    def __init__(self,name, age):
        self.__name = name
        self.__age = age

    def __call__(self):
        print(f'Name: {self.__name} - Age: {self.__age}')

    def set_name(self, name):
        self.__name = name

    def set_age(self, age):
        self.__age = age

#test: create an object
cat = Cat('Calico', 0.4)
cat()

cat.set_age(1.5)
cat()
```

The main difference here is Encapsulation (specifically, private attributes vs. public attributes).

1. First Screenshot: Public Attributes (self.age)
- In the first one, name and age are public: You can access and change them directly from outside the class.
- The downside: Anyone using your code can accidentally assign bad data to it, like cat.age = -100 or cat.age = "hello", and Python won't stop them.


2. Second Screenshot: Private Attributes (self.__age)
Notice the double underscores __ before the variable names (self.__name, self.__age).
- __ makes variables private: Python hides these attributes so they cannot be modified directly from outside using cat.__age = 1.5.
- Setter methods (set_age): To change a private attribute, you are forced to use a method like cat.set_age(1.5).

Why use private attributes (__) and setter methods?
- Using setters gives you control and validation.
- In the second example, if someone passes a negative age, you can block it inside the method
- If age were public (first screenshot), you couldn't prevent someone from setting cat.age = -5. By making it private with __ (second screenshot), you force changes to go through set_age(), keeping your object's data safe and valid!

```python
def set_age(self, age):
    if age < 0:
        print("Age cannot be negative!")
    else:
        self.__age = age
```

# **2. INHERITANCE MATH**

**INHERIT FROM THE SUPERCLASS**

```python
class Math1:
    def is_even(self, number):
        if number %2:
            return False
        else:
            return True

    def factorial(self, number):
        result = 1

        for i in range(1, number + 1):
            result *= i

        return result
```

```python
#test Math1

math1 = Math1()

#isEven() sample: number = 5 -> False
#isEven() sample: number = 6 -> True
print(math1.is_even(5))
print(math1.is_even(6))

#factorial() sample: number=4 -> 24
#factorial() sample: number=5 -> 120
print(math1.factorial(4))
print(math1.factorial(5))
```

```python
class Math2:
    def is_even(self, number):
        if number %2:
            return False
        else:
            return True

    def factorial(self,number):
        result = 1

        for i in range(1, number + 1):
            result *= i

        return result

    def estimate_euler(self, number):
        result = 1

        for i in range(1, number + 1):
            result = result + 1/self.factorial(i)

        return result
```

```python
#test Math2
math2 = Math2()

# isEven() sample: number=5 -> False
# isEven() sample: number=6 -> True
print(math2.is_even(5))
print(math2.is_even(6))

# factorial() sample: number=4 -> 24
# factorial() sample: number=5 -> 120
print(math2.factorial(4))
print(math2.factorial(5))

# estimateEuler() sample: number=2 -> 2.5
# estimateEuler() sample: number=8 -> 2.71
print(math2.estimate_euler(2))
print(math2.estimate_euler(8))
```

**OPEN FOR EXTENSION**

```python
from abc import ABC, abstractmethod

#Shape
class Shape(ABC):
    @abstractmethod
    def compute_area(self):
        pass

#Square
class Square(Shape):
    def __init__(self, side):
        self.__side = side

    def compute_area(self):
        return self.__side**2

#circle
class Circle(Shape):
    def __init__(self, radius):
        self.__radius = radius

    def compute_area(self):
        return (self.__radius**2)*3.14

#Canva
class Canva:
    def draw(self, a_shape):
        return a_shape.compute_area()
```

```python
#test 
canva = Canva()
print(canva.draw(Circle(1)))
print(canva.draw(Square(5)))
```

```python
#Rectangle

class Rectangle(Shape):
    def __init__(self, width, height):
        self.__width = width
        self.__height = height

    def compute_area(self):
        return self.__width * self.__height

#test
canva = Canva()
print(canva.draw(Circle(1)))
print(canva.draw(Square(5)))
print(canva.draw(Rectangle(4,6)))
```

# **6. CUSTOM CLASS - PYTORCH**

**SIGMOID**

```python
import torch
import torch.nn as nn 

sigmoid_function = nn.Sigmoid()
data = torch.Tensor([1,2,3])
output = sigmoid_function(data)
print(output)
```

```python
class MySigmoid(nn.Module):
    def __init__(self):
        super().__init__()

    def forward(self, x):
        return 1 / (1 + torch.exp(-x))
```

```python
my_sigmoid = MySigmoid()
output = my_sigmoid(data)
print(output)
```

**RELU**

```python
import torch
import torch.nn as nn

class MyReluActivation(nn.Module):
    def __init__(self):
        super().__init__()

    def forward(self, x):
        return torch.clamp(x, min = 0)
```

```python
data = torch.Tensor([1, -2, 3])
my_relu = MyReluActivation()
output = my_relu(data)
print(output)
```
