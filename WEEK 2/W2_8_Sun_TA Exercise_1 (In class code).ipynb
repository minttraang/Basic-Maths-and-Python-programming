# **DATA STRUCTURE**

```python
# create a list

nums = [12, 23, 60, 7, 7]
print(nums)

# acesses elements
print(nums[0])
print(nums[1:3])
```

```python
#update elements
nums = [12, 23, 60]
print(nums)

nums[0] = 20
print(nums)

for item in nums:
    print(item)
```

```python
#update elements
nums = [12, 23, 60]
print(nums)

nums.append(50) # thêm 1 phần tử vào cuối
print(nums)

nums.extend([1, -1]) # nối thêm nhiều phần tử
print(nums)

'''
append() adds a single item (of any type) to the end of a list.
extend() adds all elements from an iterable (like a list, tuple, or set) to the end of the current list.

append() method adds a single element to the end of a list. This element can be a number, string, list, or any object.

extend() method adds all elements from an iterable to the end of the current list. 
Unlike append(), it does not add the iterable as a single element; instead, each element is added individually.
'''

nums.remove(12)  # xoá giá trị 12
print(nums)

nums.clear() # xoá toàn bộ
print(nums)
```

**OTHER LIST METHODS**

```python
nums = [10,20,30]
nums.insert(1,15)  # chèn 15 vào vị trí 1
print(nums)

print(nums.index(30))  # vị trí đầu tiên của 30

print(nums.count(20)) # số lần 20 xuất hiện

nums.sort() # sắp xếp tăng dần
print(nums)

nums.reverse() # đảo ngược
print(nums)

backup = nums.copy()  # tạo bản sao
print(nums)
```

```python
# create a tuple
t = (1, 3, -2.0, "hello")
print(t)

#acess elements
print(t[0])
print(t[-1])
print(t[2:4])
```

```python
#iterate througha tuple

t = (1, -2.0, "hello")

for item in t:
    print(item)

t[2] = "hi" # t[2] = "Hi"  # Lỗi: tuple không thể thay đổi
```

```python
#create a set
s = {1,2,3,1}
print(s)

#add an element
s.add("hello")
print(s)

for item in s:
    print(item)
```

```python
s = {1,2,3,1}
print(s)

print(all(s))
'''
Checks if every single element in the iterable is considered "truthy" (i.e., evaluates to True in Python).
In Python, any non-zero number is True, and 0 is False.
Since 1, 2, and 3 are all non-zero, all(s) returns True.
'''
print(any(s))
'''
Checks if at least one element is "truthy".
Since all three numbers are truthy, any(s) returns True.
'''
```

```python
s = {1,2,3,1}
print(s)
s.add("Hello")
print(s)

s.update({1})
print(s)

print(s.issuperset({2,3}))

s.remove(2)
print(s)

s.discard(3)
print(s)

s.clear()
print(s)

'''
How they handle missing elements: remove() raises a KeyError if the element does not exist in the set, while discard() does nothing and fails silently. 
Both methods behave exactly the same way when the element is present, removing it from the set successfully.
'''
```

**Other method of Set: union, intersection, issuperset, update, discard, remove, pop, copy, clear**

```python
a = {1, 2, 3}
b = {3, 4, 5}
print(a.union(b))         # hợp
print(a.intersection(b))  # giao
print(a.issubset(b))      # con?
print(a.issuperset({1}))  # chứa?
a.update({6})    # thêm nhiều
a.discard(9)     # xoá nếu có
a.remove(1)      # xoá
print(a.pop())   # lấy ra 1 phần tử
c = a.copy()     # bản sao
a.clear()        # xoá hết
print(a)
```

**Other method of dictionary: get, pop, popitem, copy, clea**

```python
d = {"name": "An", "age": 20}
print(d.get("name"))      # "An"
print(d.get("city", ""))  # mặc định nếu thiếu
print(d.pop("age"))       # xoá & trả về value
d["city"] = "Hanoi"
print(d.popitem())        # xoá & trả về cặp cuối
backup = d.copy()         # bản sao
d.clear()                 # xoá toàn bộ
print(d)
```
