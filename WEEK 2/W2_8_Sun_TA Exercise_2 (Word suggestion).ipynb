## 1. Getting Max Over Kernel

Cho `num_list` và cửa sổ trượt kích thước `k` (k ≥ 1) di chuyển từ trái sang phải, trả về danh sách giá trị lớn nhất trong mỗi cửa sổ.

- Input: `num_list = [3, 4, 5, 1, -44, 5, 10, 12, 33, 1]`, `k = 3`
- Output: `[5, 5, 5, 5, 10, 12, 33, 33]`

```python
def max_kernel(num_list: list, k: int) -> list:
    '''
    Trả về danh sách các giá trị lớn nhất trong mỗi cửa sổ con (window) kích thước k chạy trượt trên danh sách num_list.

    Parameters:
        num_list (list): Danh sách các số đầu vào.
        k (int): Kích thước của cửa sổ trượt.

    Returns:
        list: Danh sách các giá trị lớn nhất trong từng cửa sổ con liên tiếp.
    '''

    star_indices = list(range(0, len(num_list) - k + 1))
    end_indices = list(range(k, len(num_list) + 1))

    results = []

    for start_index, end_index in zip(star_indices, end_indices):
        sub_list = num_list[start_index:end_index]
        results.append(max(sub_list))

    return results

#testing
assert max_kernel([3, 4, 5, 1, -44], 3) == [5, 5, 5]
print(max_kernel([3, 4, 5, 1, -44, 5, 10, 12, 33, 1], 3))
```

## 2. Character Counting

Đếm số lần xuất hiện của từng ký tự trong một từ, trả về dictionary `{ký_tự: số_lần}`. Chuỗi được chuyển về chữ thường trước khi đếm (không phân biệt hoa/thường).

- Input: `"smiles"` → Output: `{"s": 2, "m": 1, "i": 1, "l": 1, "e": 1}`

```python
def count_char(word: str) -> dict:

    '''
    Count the time a character appear in givens string.
    String will be converted to lowercase before counting (e.g. 'B' and 'b' is the same)

        Parameters:
            word (str): the string that needs to be analysed

        Returns:
            dict: the dictionary with each char is a key, value is the number of occurrence
    '''

    char_statistic = {}

    for chars in word.lower():
        if chars in char_statistic:
            char_statistic[chars] += 1
        else:
            char_statistic[chars] = 1

    return char_statistic


#testing
assert count_char("Baby") == {'b': 2, 'a': 1, 'y': 1}
print(count_char('smiles'))

```

## 3. Word Counting

Đọc nội dung file txt, đếm số lần xuất hiện của từng từ, trả về dictionary `{từ: số_lần}`. Tiền xử lý: chuyển chữ thường, bỏ dấu `.` và `,`, tách từ.

**Từ bài này dùng type hint đầy đủ:** `dict[str, int]` nghĩa là key kiểu `str`, value kiểu `int`; `list[str]` là list các phần tử `str`.

```python
def preprocess_text(sentence: str) -> list[str]:
    '''
    Tiền xử lý một câu: chuyển chữ thường, bỏ dấu chấm (.) và phẩy (,), rồi tách thành danh sách từ.

    Parameters:
        sentence (str): Chuỗi câu đầu vào.

    Returns:
        list[str]: Danh sách các từ sau khi xử lý.
    '''

    sentence = sentence.lower()
    sentence = sentence.replace('.','').replace(',','')
    return sentence.split()

def count_word(file_path: str) -> dict[str, int]:
    '''
    Đọc nội dung file txt và đếm số lần xuất hiện của từng từ sau khi tiền xử lý.

    Parameters:
        file_path (str): Đường dẫn đến file.

    Returns:
        dict[str, int]: Từ điển với key là từ (str), value là số lần xuất hiện (int).
    '''

    with open(file_path, 'r', encoding='utf-8') as f:
        document = f.read()
    words = preprocess_text(document)

    counter = {}
    for word in words:
        if word in counter:
            counter[word] += 1
        else:
            counter[word] = 1
    return counter


#testing (need file./P1_data.txt)
expected = ['i','love','ai','ai','is','not','easy']
assert preprocess_text('I love AI. AI is not easy') == expected

result = count_word('./P1_data.txt') 
assert result.get('who',0) == 3
print(result.get('man',0))
print(result.get('programming',0))
```

Ý nghĩa của số , 0 đằng sau là giá trị mặc định (default value) mà phương thức .get() sẽ trả về nếu từ đó không có trong từ điển (result).

Cách .get() hoạt động
Thông thường, khi truy xuất dữ liệu từ một từ điển (dict) trong Python:

    Nếu dùng cách thông thường result['man']: Nếu từ 'man' không tồn tại trong file data.txt, Python sẽ lập tức báo lỗi KeyError và dừng chương trình.

    Nếu dùng cú pháp dict.get(key, default):

    Nếu key có trong dictionary: Hàm trả về số lần xuất hiện của từ đó.

    Nếu key không có trong dictionary: Hàm sẽ trả về giá trị default (ở đây là 0) thay vì báo lỗi.

## 4. Levenshtein Distance

Tính khoảng cách chỉnh sửa tối thiểu giữa hai chuỗi (số phép chèn/xóa/thay thế ít nhất để biến `token1` thành `token2`), dùng quy hoạch động.

- Ví dụ: `levenshtein_distance("kitten", "sitting")` → `3`

```python
def levenshtein_distance(token1: str, token2: str) -> int:
    '''
    Tính khoảng cách Levenshtein (edit distance) giữa hai chuỗi bằng quy hoạch động.

    Parameters:
        token1 (str): Chuỗi nguồn.
        token2 (str): Chuỗi đích.

    Returns:
        int: Số phép chỉnh sửa tối thiểu (chèn, xóa, thay thế).
    '''

    distance = [[0] * (len(token2) + 1) for _ in range(len(token1) + 1)]

    for t1 in range(len(token1) + 1):
        distance[t1][0] = t1
    for t2 in range(len(token2) + 1):
        distance[0][t2] = t2

        for t1 in range(1, len(token1) + 1):
            for t2 in range(1, len(token2) + 1):
                if token1[t1-1] == token2[t2-1]:
                    distance[t1][t2] = distance[t1-1][t2-1]
                else:
                    insert_cost = distance[t1][t2-1]
                    delete_cost = distance[t1-1][t2]
                    substitute_cost = distance[t1-1][t2-1]
                    distance[t1][t2] = min(insert_cost, delete_cost, substitute_cost) + 1

    return distance[len(token1)][len(token2)]

#testing
assert levenshtein_distance("hi", "hello") == 4
assert levenshtein_distance("kitten", "sitting") == 3
print(levenshtein_distance("hola", "hello"))
```
