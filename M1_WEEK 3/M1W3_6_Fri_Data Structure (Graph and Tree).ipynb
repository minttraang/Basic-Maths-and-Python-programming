# **GRAPH DATA STRUCTURE**

## 1. What is a Graph?

A **graph** is a non-linear data structure used to represent relationships between objects.

A graph consists of:

- **Vertices / Nodes**: the objects in the graph
- **Edges**: the connections between objects

For example, in a social network:

- each person can be represented as a node
- each friendship or connection can be represented as an edge

## 2. Representing a Graph in Python

A common way to represent a graph in Python is using an **adjacency list**.

In an adjacency list:

- each key is a node
- each value is a list of neighbors connected to that node

```python
graph = {
    "A": ["B","C","D"],
    "B": ["A"],
    "C": ["A", "E"],
    "D": ["A"],
    "E": ["C"]
}

graph
```

## 3. Neighbors and Degree

The **neighbors** of a node are the nodes directly connected to it.

The **degree** of a node is the number of edges connected to that node.

In an undirected graph, the degree of a node is usually equal to the number of neighbors in its adjacency list.

```python
def get_neighbors(graph, node):
    return graph[node]

def degree(graph, node):
    return len(graph[node])

node = "A"
print("Neighbors of A:", 
get_neighbors(graph, node))

print("Degree of A:", degree(graph, node))
```

## 4. Undirected Graph

An **undirected graph** has edges without direction.

If A is connected to B, then B is also connected to A.

```python
undirected_graph = {
    "A": ["B", "C"],
    'B': ['A', 'D'],
    'C': ['A', 'D', 'E'],
    'D': ['B', 'C'],
    'E': ['C']
}

print(undirected_graph)
```

## 5. Directed Graph

A **directed graph** has edges with direction.

If A points to B, it does not necessarily mean that B points back to A.

```python
directed_graph = {
    'A': ['B', 'C'],
    'B': ['D'],
    'C': ['D', 'E'],
    'D': [],
    'E': []
}

print(directed_graph)
```

## 6. Weighted Graph

A **weighted graph** has edges with weights.

A weight can represent distance, cost, time, or any other numeric value.

```python
weighted_graph = {
    'A': [('B', 4), ('C', 7)],
    'B': [('A', 4), ('D', 2)],
    'C': [('A', 7), ('D', 5),('E', 3)],
    'D': [('B', 2), ('C',5)],
    'E': [('C', 3)]
}

print(weighted_graph)
```

## 7. Traversal and Visited Set

Graph traversal means visiting nodes in a graph systematically.

Unlike a tree, a graph may contain cycles. Therefore, we often use a **visited set** to avoid visiting the same node again and again.

```python
cyclic_graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'D', 'E'],
    'D': ['B', 'C'],
    'E': ['C']
}

visited = set()

print("Initial visited set: ", visited)
```

## 8. Depth-First Search (DFS) with Recursion

**Depth-First Search (DFS)** explores as far as possible along one branch before backtracking.

DFS is often implemented using recursion. Recursion uses the program's call stack behind the scenes.

```python
def dfs_recursive(graph, node, visited=None, order=None):
    if visited is None:
        visited = set()
    if order is None:
        order = []

    if node in visited:
        return order

    visited.add(node)
    order.append(node)

    for neighbor in graph[node]:
        dfs_recursive(graph, neighbor, visited, order)
    return order

dfs_order = dfs_recursive(cyclic_graph,'A')
print("DFS traversal order:", dfs_order)
```

## 9. Depth-First Search (DFS) with Stack

DFS can also be implemented explicitly using a **Stack**.

A stack follows the **Last In, First Out (LIFO)** principle.

To process neighbors from left to right, we push them into the stack in reversed order.

```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        return self.items.pop()

    def is_empty(self):
        return len(self.items) == 0

    def display(self):
        return self.items

class Graph:
    def __init__(self, graph):
        self.graph = graph

    def dfs_stack(self, start):
        visited = {start}
        stack = Stack()
        stack.push(start)
        order = []

        while not stack.is_empty():
            node = stack.pop()
            order.append(node)

            for neighbor in reversed(self.graph[node]):
                if neighbor not in visited:
                    visited.add(neighbor)
                    stack.push(neighbor)

        return order

g = Graph(cyclic_graph)
order = g.dfs_stack('A')
print('DFS traversal order:', order)
```
